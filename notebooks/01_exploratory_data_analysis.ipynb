# 01 – Exploratory Data Analysis
**AquaML Portfolio Project**

Quick tour of the three tabular sample datasets we will use throughout the project:
- Daily streamflow + meteorology
- Water quality measurements
- Hydrological situation reports (text)

This notebook establishes data quality, distributions, and basic relationships before modeling.


```
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid", context="notebook")
pd.set_option("display.max_columns", 30)
```

## 1. Streamflow Time Series

```
sf = pd.read_csv(ROOT / "data/sample/streamflow_daily.csv", parse_dates=["date"])
print(sf.shape)
sf.head()
```

```
fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
axes[0].plot(sf["date"], sf["streamflow_m3s"], color="#1f77b4", linewidth=0.9)
axes[0].set_ylabel("Streamflow (m³/s)")
axes[0].set_title("Daily Streamflow")
axes[1].bar(sf["date"], sf["precip_mm"], color="#2ca02c", width=1.0, alpha=0.7)
axes[1].set_ylabel("Precip (mm)")
axes[2].plot(sf["date"], sf["temp_c"], color="#d62728", linewidth=0.8)
axes[2].set_ylabel("Temp (°C)")
fig.tight_layout()
plt.show()
```

```
print(sf.describe().round(2))
print("\nMissing values:")
print(sf.isna().sum())
```

## 2. Water Quality

```
wq = pd.read_csv(ROOT / "data/sample/water_quality.csv")
print(wq.shape)
print(wq["quality"].value_counts())
wq.head()
```

```
fig, axes = plt.subplots(2, 3, figsize=(13, 7))
for ax, col in zip(axes.flat, ["ph", "turbidity_ntu", "dissolved_oxygen_mgl", "bod_mgl", "nitrate_mgl", "temp_c"]):
    for q, color in zip(["Good", "Fair", "Poor"], ["#2ca02c", "#ff7f0e", "#d62728"]):
        subset = wq[wq["quality"] == q][col]
        ax.hist(subset, bins=25, alpha=0.55, label=q, color=color)
    ax.set_title(col)
    ax.legend(fontsize=8)
fig.suptitle("Feature distributions by water quality class", fontsize=14)
fig.tight_layout()
plt.show()
```

## 3. Hydrology Reports (NLP)

```
reports = pd.read_csv(ROOT / "data/sample/hydrology_reports.csv")
print(reports.shape)
print(reports["severity"].value_counts())
reports.sample(5, random_state=42)
```

```
reports["text_len"] = reports["text"].str.len()
print(reports.groupby("severity")["text_len"].describe().round(1))
```

## Summary
- Streamflow shows clear seasonality + extreme events → good candidate for forecasting.
- Water quality classes are separable by the measured parameters.
- Reports are short, labeled by severity → ideal for classification + topic exploration.
